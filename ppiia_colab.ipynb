## 5. Launch a Gradio web app (public URL)

This cell spins up an interactive UI you can share with anyone. Use the public URL to record your demo video.

```python
import gradio as gr
import json
from ppiia import analyze, llm_available

def run(pdf_file, url, raw_text):
    source = None
    if pdf_file is not None:
        source = pdf_file.name
    elif url and url.strip():
        source = url.strip()
    elif raw_text and raw_text.strip():
        source = raw_text.strip()
    else:
        return 'Please upload a PDF, paste a URL, or paste some text.', '', '', '', '', {}
    try:
        rep = analyze(source).to_dict()
    except Exception as e:
        return f'Error: {e}', '', '', '', '', {}
    title_md = f"### {rep.get('title', 'Bill')}\n\n**Ministry:** {rep.get('ministry') or '—'}"
    impacts = rep.get('impacts', {})
    impact_md = (
        f"**Short-term (0–1 yr):** {impacts.get('short_term', '—')}\n\n"
        f"**Medium-term (1–5 yr):** {impacts.get('medium_term', '—')}\n\n"
        f"**Long-term (>5 yr):** {impacts.get('long_term', '—')}"
    )
    sectors_md = '\n'.join(f'- {s}' for s in rep.get('sectors', [])) or '—'
    posneg_md = '**Positives:**\n' + ('\n'.join(f'- {p}' for p in rep.get('positives', [])) or '—') \
              + '\n\n**Negatives:**\n' + ('\n'.join(f'- {n}' for n in rep.get('negatives', [])) or '—')
    return title_md, rep.get('summary', ''), impact_md, sectors_md, posneg_md, rep

mode_badge = f"**Mode:** {llm_available().upper()} (local Hugging Face — no API keys)"

with gr.Blocks(title='PPIIA - Bill Analyzer') as demo:
    gr.Markdown(f'# 🏛️ Public Policy Insight & Impact Analyzer\n{mode_badge}')
    with gr.Row():
        pdf_in = gr.File(label='Upload Bill PDF', file_types=['.pdf'])
        url_in = gr.Textbox(label='OR paste URL')
    txt_in = gr.Textbox(label='OR paste raw text', lines=4)
    btn = gr.Button('Analyze', variant='primary')
    title_out = gr.Markdown()
    with gr.Tab('Summary'):
        summary_out = gr.Textbox(label='Citizen-friendly summary', lines=12)
    with gr.Tab('Impacts'):
        impacts_out = gr.Markdown()
    with gr.Tab('Sectors'):
        sectors_out = gr.Markdown()
    with gr.Tab('Positives / Negatives'):
        posneg_out = gr.Markdown()
    with gr.Tab('Raw JSON'):
        raw_out = gr.JSON()
    btn.click(run, [pdf_in, url_in, txt_in],
              [title_out, summary_out, impacts_out, sectors_out, posneg_out, raw_out])

demo.launch(share=True)
```
